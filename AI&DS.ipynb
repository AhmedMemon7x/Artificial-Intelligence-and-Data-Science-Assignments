```python
name = input("Enter your full name: ")
sentence = input("Enter a sentence: ")
number = input("Enter a number or alphanumeric value: ")

print("\n1. upper():", name.upper())

print("2. lower():", name.lower())

print("3. title():", sentence.title())

print("4. capitalize():", sentence.capitalize())

print("5. swapcase():", name.swapcase())

print("6. strip():", name.strip())

print("7. lstrip():", name.lstrip())

print("8. rstrip():", name.rstrip())

print("9. replace():", sentence.replace(" ", "_"))

print("10. split():", sentence.split())

words = sentence.split()
print("11. join():", "-".join(words))

print("12. find():", sentence.find("a"))

if "a" in sentence:
    print("13. index():", sentence.index("a"))
else:
    print("13. index(): Character 'a' not found")

print("14. count():", sentence.count("a"))

print("15. startswith():", sentence.startswith("Hello"))

print("16. endswith():", sentence.endswith("."))

print("17. isalpha():", name.isalpha())

print("18. isdigit():", number.isdigit())

print("19. isalnum():", number.isalnum())

print("20. isspace():", name.isspace())

print("21. islower():", name.islower())

print("22. isupper():", name.isupper())

print("23. istitle():", name.istitle())

print("24. center():", name.center(30, "*"))

print("25. ljust():", name.ljust(30, "-"))

print("26. rjust():", name.rjust(30, "-"))

print("27. zfill():", number.zfill(10))

print("28. partition():", sentence.partition(" "))

print("29. rpartition():", sentence.rpartition(" "))

print("30. casefold():", name.casefold())
```
